Data Structure<br>
Data Structure is a method to arrange data in a structured way, so it can be used efficently!

List <br>
List is an ordered, mutable collection of items.<br>
Ordered : It has some permanent position <br>
Mutable : It can be changed after creation
It can store duplicates<br>
It is declared using square brackets

```python
# Empty list
my_list = []

# List with items
numbers = [1, 2, 3, 4, 5]
names = ["Alice", "Bob", "Charlie"]
mixed = [1, "hello", 3.14, True]
print(numbers, names, mixed)
```

len() Method

```python
fruits = ["apple", "banana", "cherry"]
print(len(fruits)) #To find the length of a list using the len() function.
```

using range() method in List

```python
x = list(range(5))
print(x)
```

To check the type:

```python
numbers = [1, 2, 3, 4, 5]
print(type(numbers)) # Gives type of list
```

```python
print(numbers[0]) # Accessing the first element of the list
print(numbers[2]) # Accessing the third element of the list
print(numbers[-1]) # Accessing the last element of the list
```

```python
print(numbers[1 : 4]) # Slicing allows you to access a range of elements in the list.
print(numbers[-4:-1]) # Slicing with negative indices allows you to access a range of elements from the end of the list.
print(numbers[::2]) #step = 2, it will return every second element from the list.
print(numbers[::-1]) #step = -1, it will return the list in reverse order.
```

insert() Method

```python
numbers.insert(2, 10) # Inserts 10 at index 2
print(numbers)
```

append() Method

```python
numbers.append(8) # Appends 8 to the end of the list
print(numbers)
```

extend() Method

```python
additional_numbers = [6, 7, 8]
numbers.extend(additional_numbers) # Extends the list by appending elements from another list
print(numbers)
```

remove() Method

```python
numbers.remove(8) # Removes the first occurrence of 8 from the list
print(numbers)
```

pop() Method

```python
numbers.pop() # Removes and returns the last item in the list
print(numbers)
numbers.pop(2) # Removes and returns the item at index 2
print(numbers)
```

```python
del numbers[1] # Deletes the item at index 1
print(numbers)
del numbers[1:3] # Deletes a slice of items from index 1 to 2
print(numbers)
del numbers # Deletes the entire list
print(numbers) # This will raise an error since the list no longer exists
```

clear() Method

```python
additional_numbers.clear() # Clears all items from the list but keeps the list itself
print(additional_numbers) # This will print an empty list: []
```

```python
fruits = [ "banana", "cherry","apple"]
for x in fruits:
    print(x)
```

Loops - for

```python
for i in range(len(fruits)):
    print(fruits[i])
```

Loops - While

```python
i = 0
while i < len(fruits):
    print(fruits[i])
    i += 1
```

List Comprehension:  <br>A list comprehension is a concise way to create a new list from an iterable.

```python
x = [i for i in fruits] # List comprehension to print each fruit
print(x)
```

extend() Method

```python
x = list(("orange", "kiwi", "mango"))
fruits.extend(x) # Appends multiple items to the end of the list
print(fruits)
```

```python

new_fruits = []
for x in fruits:
    if "a" in x:
        new_fruits.append(x)
print(new_fruits)
```

reverse() Method

```python
fruits.reverse()# Reverses the order of the list
print(fruits)
```

sort() Method

```python
fruits.sort() # Sorts the list in ascending order
print(fruits)
```

```python
# reverse() will reverse the order of the list, whereas sort(reverse=True) will sort the list in descending order.
fruits.sort(reverse=True) # Sorts the list in descending order
print(fruits)
```
