Join set<br>
union() and update() combine all unique elements from both sets into one set.<br>
intersection() returns only the elements that are present in both sets.<br>
difference() returns the elements that exist in the first set but not in the other set(s).<br>
symmetric_difference() returns the elements that are unique to each set, excluding the common elements.<br>

```python
set1= {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.union(set2)  #set3 = set1 | set2 both are same
print("Union:", set3)

set4 = set1.intersection(set2) #set4 = set1 & set2 both are same
print("\nIntersection:", set4)

set5 = set1.difference(set2) #set5 = set1 - set2 both are same
print(f"\nDifference: {set5} \n Set1: {set1} \n Set2: {set2}\n")

set6 = set1.symmetric_difference(set2) #set6 = set1 ^ set2 both are same
print("Symmetric Difference:", set6)

set7 = set1.difference_update(set2) #set7 = set1 -= set2 both are same
print(f"\nDifference Update: {set7} \n Set1: {set1} \n Set2: {set2}")
```

Difference between union() and update(): <br>
union() will keep the original set and produce new set of items but update will modify the current set<br>
difference_update() method removes the items that are present in both sets.<br>
The difference_update() method is different from the difference() method, because the difference() method returns a new set, without the unwanted items, and the difference_update() method removes the unwanted items from the original set.
