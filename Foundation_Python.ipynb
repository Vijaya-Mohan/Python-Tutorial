Calculate Average
Write a program that calculates the average of numbers in a list without predifined function.

Input:

scores = [85, 92, 78, 95, 88]
Expected Output:

Average score: 87.60

```python
scores =[85, 92, 78, 95, 88]
total = 0
for i in scores:
    total = total + i
avg = total / len(scores)
print(f"Scores\t: {scores} \nTotal\t: {total} \nAverage\t: {avg}")
```

Count Vowels in Names
Given a list of names, count how many names start with a vowel (A, E, I, O, U).

Input:

names = ["Alice", "Bob", "Emily", "David", "Oliver", "Sarah"]
Expected Output:

Names starting with vowels: 3

```python
names = ["Alice", "Bob", "Emily", "David", "Oliver", "Sarah"] 
count = 0 
for i in names: 
    if i.startswith(("A", "E", "I", "O", "U")): 
        count = count + 1 
print("Names starting with vowels:", count)
```

Find Maximum and Its Position
Find both the maximum value in a list and its index position.

Input:

numbers = [45, 23, 67, 89, 12, 56]
Expected Output:

Maximum value: 89

```python
numbers = [45, 23, 67, 89, 12, 56] 
max = 0 
for i in numbers: 
    if max < i: max = i 
    index = numbers.index(i) 
print(f"Maximum value: {max}\nPosition: {index} ")
```

Find All Occurrences
Find all index positions where a specific value appears in a list.

Input:

numbers = [5, 2, 8, 2, 9, 2, 1, 2]
target = 2
Expected Output:

Value 2 found at positions: [1, 3, 5, 7]

```python
numbers = [5, 2, 8, 2, 9, 2, 1, 2] 
target = 2 
a = [] 
for i in range(len(numbers)): 
    if numbers[i] == target: 
        a.append(i) 
print("Value 2 found at positions:", a)
```
