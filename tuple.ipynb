Tuple : <br>
Tuple is ordered immutable collection of elements.<br>
Ordered & Indexed: Items have a fixed position, and you can access them using zero-based integer indices<br>
Immutable : Unlike lists, you cannot change, add, or remove items after a tuple has been created. 
<br>This makes data tamper-proof and allows programs to run faster and consume less memory.<br>



```python
the_tuple = (1, )
the_tuple2 = (1, 2, 3)
print(type(the_tuple), type(the_tuple2))  # Both are tuples
x = [1, 2, 3]
y= tuple(x)  # Convert list to tuple
print(type(x), type(y))  # x is a list, y is a tuple
```

len() Method  

```python
print(len(the_tuple))  
```

Accessing Tuple

```python
print(the_tuple2[1]) #Accessing the second element of the tuple
print(the_tuple2[-1]) #Accessing using negative indexing
print(the_tuple2[1:])  #Accessing elements from index 1 to the end of the tuple
print(the_tuple2[:2])  #Accessing elements from the start of the tuple to index 2
print(the_tuple2[1:3])  #Accessing a range of elements in the tuple
print(the_tuple2[-3:-1])  #Accessing a range of elements using negative indexing
```

Concatenating tuple<br>
As you cannot append items so the only way to add items is concate.

```python
new_tuple = the_tuple + the_tuple2  # Concatenating two tuples
print(new_tuple)  #Tuple are ordered and allows duplicates
```

To remove an item from tuple is impossible as they are immutable, only way is to convert to 'List', remove item(s), convert back to tuple.

```python
print(new_tuple)
new_list = list(new_tuple)  # Convert tuple to list
new_list.remove(1)  # Remove the first occurrence of 1 from the list
new_tuple = tuple(new_list)  # Convert the list back to a tuple
print(new_tuple)  #Print the modified tuple
```

Here creating tuple means we are packing items, python also providing unpacking items.

```python
print(new_tuple)
```

```python
alpha, beta, gamma = new_tuple  # Unpacking the tuple into variables
print(alpha, beta, gamma)  # Print the unpacked values
```

Also if don't know the exact count to un-pack, we can go for * 

```python
*alpha, beta = new_tuple #
print(alpha, beta)  #Print the unpacked values with the first two elements in a list and the last element in a variable
```

Looping through tuple using 'for'

```python
for i in new_tuple:
    print(i)  #Iterating through the tuple and printing each element
```

```python
#Loop Through the Index Numbers
for i in range(len(new_tuple)):
    print(new_tuple[i])
```

Looping through tuple using 'while' loop

```python
i = 0
while i < len(new_tuple):
    print(new_tuple[i])
    i +=1
```

```python
print(new_tuple * 2)
```
