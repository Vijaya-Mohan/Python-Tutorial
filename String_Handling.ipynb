Python typically don't have array, instead it will take a single character as a String with length as 1<br>
 

```python
#Looping through a String
a = "Welcome"
for x in a:
    print(x)
```

```python
#To find length of the String
print(len(a))
```

```python
#To check certain phrase or sting in a string using 'in'
a = "String Handling"
print("String" in a)
print("string" in a) #Note python are case sensitive
print("String" not in a) #To check certain phrase or sting is not present in a string using 'not in'
```

Python strings are immutable, you can acess using square brackets[]. You can access characters, but you cannot alter them directly

```python
a = 'Hello Python'
print(a[0])
print(a[-1]) #Negative Indexing also allowed

```

```python
#But accessing out of Range is not allowed, will raise IndexError
print(a[50])
```

Slicing String

```python
a = 'Hello Python'
print(a[2:]) #Starts from index 2 to end
print(a[:5]) #Starts from index 0 to 4
print(a[2:5]) #Starts from index 2 to 4
print(a[-5:-2]) #Starts from index -5 to -3 - Negative indexing also allowed
```

```python
a = 'Hello Python'
print(a.upper()) #Converts all characters to Uppercase
print(a.lower()) #Converts all characters to Lowercase
print(a.strip()) #Removes any whitespace from the beginning or the end
print(a.replace('Hello', 'hello')) #Replaces a string with another string
print(a.split()) #Splits the string into a list
```

String Concatenation

```python
a = "Hello"
b = "Python"
print(a + b) #Concatenation of two strings
```

Format String - F String

```python
Celsius = 37.5
print(f"The temperature is {Celsius} Celsius") #f-string formatting
```
